```python
def my_tee(iterable, n):
    values = list(iterable)
    return [values] * n
```

```python
squares = (n ** 2 for n in range(3))
sq1, sq2 = my_tee(squares, 2)
```

```python
for element in sq1:
    print(element)
```

```python
for element in sq2:
    print(element)
```

```python
def my_tee2(iterable, n):
    values = list(iterable)
    lists = [values] * n
    iterators = [iter(each_list) for each_list in lists]
    return iterators

squares = (n ** 2 for n in range(3))
sq1, sq2 = my_tee2(squares, 2)
```

```python
for element in sq1:
    print(element)
```

```python
for element in sq1:
    print(element)
```

```python
from collections import deque

def my_tee(iterable, n):
    iterator = iter(iterable)
    tee_objects = [_my_tee(iterator) for _ in range(n)]
    for obj in tee_objects:
        obj.siblings = tee_objects
    return tuple(tee_objects)

class _my_tee:  # Needs to be an iterator...
    def __init__(self, source_iterable):
        self.source = source_iterable
        self.buffer = deque()
        self.siblings = []

    def __iter__(self):  # matches the built-in `iter`
        return self

    def __next__(self):  # this is the fun one
        if not self.buffer:
            value = next(self.source)
            for obj in self.siblings:
                obj.buffer.append(value)
        return self.buffer.popleft()
```

```python
def my_pairwise(iterable):
    it1, it2 = my_tee(iterable, 2)
    next(it2)
    return zip(it1, it2)
```

```python
for before, after in my_pairwise("Hello from Lisbon".split()):
    print(before, after)
```
