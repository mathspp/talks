```python
from itertools import pairwise
```

```python
countries = ["Estonia", "Latvia", "Lithuania"]

for north, south in pairwise(countries):
    print(f"{north} is North of {south}.")
```

```python
def my_pairwise(iterable):
    i1, i2 = tee(iterable, 2)
    next(i2)
    yield from zip(i1, i2)
```

```python
for north, south in my_pairwise(countries):
    print(f"{north} is North of {south}.")
```

```python
from collections import deque

def my_tee(iterable, n):
    iterator = iter(iterable)
    objects = [_tee(iterator) for _ in range(n)]
    for obj in objects:
        obj.siblings = objects
    return tuple(objects)

class _tee:
    def __init__(self, source):
        self.source = source
        self.buffer = deque()
        self.siblings = None

    def __iter__(self):
        return self

    def __next__(self):
        if not self.buffer:
            element = next(self.source)
            for sibling in self.siblings:
                sibling.buffer.append(element)
        return self.buffer.popleft()
```

```python
squares = (x ** 2 for x in range(3))
copies = my_tee(squares, 44)
for copy in copies:
    print(list(copy))
```
